```python
import pandas as pd
import os
import glob

# lokasi folder dataset
DATA_DIR = "."

# mencari seluruh file csv termasuk subfolder
csv_files = glob.glob(
    os.path.join(DATA_DIR, "**", "*.csv"),
    recursive=True
)

print("Jumlah CSV:", len(csv_files))
```

```python
dataframes = {}

for file in csv_files:

    # mengambil nama file tanpa ekstensi
    name = os.path.splitext(os.path.basename(file))[0]

    # membaca csv
    df = pd.read_csv(file)

    # menyimpan dataframe
    dataframes[name] = df

    print("\n" + "="*60)
    print(f"FILE: {name}")
    print("="*60)

    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())
```

```python
# melihat seluruh data dictionary
dictionary = dataframes["DATA_DICTIONARY"]

pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", None)

display(
    dictionary.sort_values(["file", "column"])
)
```

```python
# memeriksa keunikan ID tabel utama

for table in ["patients", "index_encounters", "train", "test"]:

    df = dataframes[table]

    print(f"\n{table.upper()}")

    for col in ["patient_id", "encounter_id", "sample_id"]:

        if col in df.columns:
            print(
                f"{col}:",
                "Unique =", df[col].nunique(),
                "| Duplicate =", df[col].duplicated().sum()
            )
```

```python
# mengambil daftar patient_id dari tabel utama

index_ids = set(dataframes["index_encounters"]["patient_id"])

patient_ids = set(dataframes["patients"]["patient_id"])

train_ids = set(dataframes["train"]["patient_id"])

test_ids = set(dataframes["test"]["patient_id"])

print("Index vs Patients:", len(index_ids & patient_ids))

print("Train vs Index:", len(train_ids & index_ids))

print("Test vs Index:", len(test_ids & index_ids))

print("Train-Test Overlap:", len(train_ids & test_ids))
```

```python
# mengubah timestamp menjadi datetime

index = dataframes["index_encounters"].copy()

index["prediction_time"] = pd.to_datetime(
    index["prediction_time"],
    errors="coerce"
)

obs = dataframes["observations"].copy()

obs["observed_at"] = pd.to_datetime(
    obs["observed_at"],
    errors="coerce"
)

# menghubungkan observation dengan prediction time
obs = obs.merge(
    index[["patient_id", "prediction_time"]],
    on="patient_id",
    how="left",
    validate="many_to_one"
)

# memeriksa data setelah prediction time
after_cutoff = obs[
    obs["observed_at"] > obs["prediction_time"]
]

print("Total observations:", len(obs))
print("After cutoff:", len(after_cutoff))
print("Before/equal cutoff:", len(obs) - len(after_cutoff))
```
