```python
# Pilih kolom tertentu saja (misalnya yang bertipe numeric)
numeric_cols = train.select_dtypes(include=['int64', 'float64']).columns

corr_matrix_subset = train[numeric_cols].corr()
print(corr_matrix_subset.head(10))  # Cek 10 baris pertama
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Gambar heatmap korelasi
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix_subset, annot=True, cmap='coolwarm', center=0)
plt.title('Matriks Korelasi Fitur')
plt.show()
```
